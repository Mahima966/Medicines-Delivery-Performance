```python
#This case study delves into a detailed analysis of order fulfillment and delivery performance for a delivery-based business that relies on efficient
#logistics to ensure timely and accurate delivery of products to customers. The business in question operates in a highly competitive and customer-centric
#environment, where the ability to meet delivery expectations significantly influences customer satisfaction and operational efficiency.

#The dataset used in this case study contains a wide range of variables that are critical to understanding the performance of the delivery process.
#These variables include key order metrics, delivery status, city-specific delivery performance, time-to-delivery (TAT) calculations, and service-level adherence (SLA)
#metrics. Each field provides valuable insights into how the company’s delivery system performs in terms of order fulfillment, delivery timeliness, and customer satisfaction.
```

```python
# Business view of data provided

#Order Creation: The process begins when an order is created in the system.
#Verification: Since this is related to medicine delivery, the first step involves the verification of the prescription. The order is sent to the verification team to ensure the prescription is accurate and valid.
#Store Processing: Once the order is verified, it is forwarded to the store for fulfillment.
#Billing: After reaching the store, the order is billed and prepared for delivery.
#Rider Assignment: Once the order is billed, a rider is assigned for delivery to the customer.
#Delivery: The rider proceeds with delivering the order to the customer’s address.
#Order Cancellation: After delivery, the customer has the option to cancel the order. If the order is cancelled, the return process may begin.
```

```python
#What we need to do

#Track the total number of orders received, delivered, cancelled, pending, and post-delivery returns.
#Track the first TAT, second TAT, and latest TAT.
#Track the time for each leg of the process:
  #Billing time
  #Pickup time
  #Rider reach store time
```

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Sample data.csv')
df
```

```python
#Created to track how many orders were attempted before getting delivered.
df['Order_Reattempt']=np.where(df['delivery_attempted'].notna(),1,0)
df
```

```python
#Created to check how many orders were delivered directly without getting cancelled after delivery
df['isDelivered']=np.where(df['LATESTORDERSTATUS']=='DELIVERED',1,0)
df
```

```python
# Created to track how many orders were cancelled
df['isCancelled']=np.where(df['pendencyFlag']=='cancelled',1,0)
df
```

```python
#Created to track orders which are pending
df['Pending']=np.where(df['pendencyFlag']=='pending',1,0)
df
```

```python
#Created to track the orders asking to be returned after delivery
df['Post_delivery_returns']=np.where(df['pendencyFlag']=='post-Delivery_Returns',1,0)
df
```

```python
#Created to track the First TAT Adherence
df['Formatted_createddate']=pd.to_datetime(df['created_ts_IST']).dt.strftime('%Y-%m-%d %H:%M:%S')
df['Formatted_createddate']=pd.to_datetime(df['Formatted_createddate'])
df['firstTAT']=pd.to_datetime(df['firstTAT']).dt.strftime('%Y-%m-%d %H:%M:%S')
df['firstTAT']=pd.to_datetime(df['firstTAT'])
```

```python
#Created to track the difference between First TAT and created date
df['Created_to_FTAT_Diff']=np.where(df['firstTAT'].notna() & df['Formatted_createddate'].notna(),
np.maximum((df['firstTAT'] - df['Formatted_createddate']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
#Created the bucket for First TAT and Created TAT difference
df['Created_to_F_TAT_Diff_0_to_2_hrs']=np.where(df['Created_to_FTAT_Diff']<121,1,0)
df
```

```python
#Created the bucket for First TAT and Created TAT difference
df['Created_to_F_TAT_Diff_2_to_4_hrs']=np.where((df['Created_to_FTAT_Diff']>120) & (df['Created_to_FTAT_Diff']<241),1,0)
df
```

```python
#Created the bucket for First TAT and Created TAT difference
df['four_8hrs']=np.where((df['Created_to_FTAT_Diff']>=241) & (df['Created_to_FTAT_Diff']<=480),1,0)
df
```

```python
#Created the bucket for First TAT and Created TAT difference
df['8_to_24hr']=np.where((df['Created_to_FTAT_Diff']>=481) & (df['Created_to_FTAT_Diff']<=1440),1,0)
df
```

```python
#Created the bucket for First TAT and Created TAT difference
df['twentyfour_48hrs']=np.where((df['Created_to_FTAT_Diff']>=1441) & (df['Created_to_FTAT_Diff']<=2880),1,0)
df
```

```python
df['grater_48hrs']=np.where(df['Created_to_FTAT_Diff']>2880,1,0)
df
```

```python
#Created to track the Orders whose terminal status is delivered and cancelled after delivery
df['NetDelivery']=np.where (df['delivered'].notna(),1,0)
df
```

```python
df['delivered']=pd.to_datetime(df['delivered']).dt.strftime('%Y-%m-%d %H:%M:%S')

df['firstTAT']=pd.to_datetime(df['firstTAT']).dt.strftime('%Y-%m-%d %H:%M:%S')
```

```python
#Created to track First TAT Adherence
df['First_SLA_Adherence'] = np.where(df['NetDelivery'] == 1, np.where(df['delivered'] <= df['firstTAT'], 1,0),np.nan)
df
```

```python

df['secondTAT']=pd.to_datetime(df['secondTAT']).dt.strftime('%Y-%m-%d %H:%M:%S')
df['latestTAT']=pd.to_datetime(df['latestTAT']).dt.strftime('%Y-%m-%d %H:%M:%S')
```

```python
##Created to track Second TAT Adherence
df['Second_SLA_Adherence'] = np.where(df['NetDelivery'] == 1, np.where(df['delivered'] <= df['secondTAT'], 1,0),np.nan)
df
```

```python
##Created to track Latest TAT Adherence
df['latest_SLA_Adherence'] = np.where(df['NetDelivery'] == 1, np.where(df['delivered'] <= df['latestTAT'], 1,0),np.nan)
df
```

```python
#Created to track total orders count
df['count']=np.where(df['orderid'].notna(),1,0)
df
```

```python
#Cretaed to track where the actual pendency is
df['prebill']=np.where((df['pendencyFlag']=='pending') & (df['pendencyType']=='pre-Billing'),1,0)
df
```

```python
#Cretaed to track where the actual pendency is
df['pre_store_pending']=np.where((df['pendencyFlag']=='pending') &( df['pendencyType']=='pre-Verification'),1,0)
df
```

```python
#Cretaed to track where the actual pendency is
df['postShip_Pending']=np.where((df['pendencyFlag']=='pending') &( df['pendencyType']=='in-Transit'),1,0)
df
```

```python
df['post_billing']=np.where((df['pendencyFlag']=='Attempted') &( df['pendencyType']=='post-Billing'),1,0)
df
```

```python
#Cretaed to track where the actual pendency is
df['DSP_Assigned']=np.where(df['forwardDSP'].notna(),1,0)
df
```

```python
#Cretaed to track whether the order is landed on store or not
df['Store_Pushed']=np.where(df['SHOPID'].notna(),1,0)
df
```

```python
df['last_shc']=pd.to_datetime(df['last_shc'])
df['billed']=pd.to_datetime(df['billed'])
```

```python
#Cretaed to track how much time order took for billing once it landed on store
df['Billed_time_min']= np.where(df['billed'].notna() & df['last_shc'].notna(),
np.maximum((df['billed'] - df['last_shc']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
df['firstpacked']=pd.to_datetime(df['firstpacked'])
df['billed']=pd.to_datetime(df['billed'])
```

```python
#Cretaed to track how much time order took for rider assignment
df['DSP_Allocation_In_Min'] = np.where(df['billed'].notna() & df['firstpacked'].notna(),
np.maximum((df['firstpacked'] - df['billed']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
df['firstshipped']=pd.to_datetime(df['firstshipped'])
df['first_rts']=pd.to_datetime(df['first_rts'])
```

```python
#Cretaed to track how much time order took to get picked up by rider
df['pickup_in_minuts'] = np.where(df['firstshipped'].notna() & df['first_rts'].notna(),
np.maximum((df['firstshipped'] - df['first_rts']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
df['delivered']=pd.to_datetime(df['delivered'])
df['lastshipped']=pd.to_datetime(df['lastshipped'])
```

```python
#Cretaed to track how much time rider took to deliver the order once it gets in hand
df['Rider_delivery_in_Min']=np.where(df['delivered'].notna() & df['lastshipped'].notna(),
np.maximum((df['delivered'] - df['lastshipped']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
#Cretaed to track how much time order took to get delivered from getting billed to delivered
df['O2D_in_min']=np.where(df['delivered'].notna() & df['billed'].notna(),
np.maximum((df['delivered'] - df['billed']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
df['rider_reached_store']=pd.to_datetime(df['rider_reached_store'])
df['first_ofd']=pd.to_datetime(df['first_ofd'])
```

```python
#Cretaed to track how much time rider took to reach the store
df['Rider_delivery_in_Min']=np.where(df['rider_reached_store'].notna() & df['first_ofd'].notna(),
np.maximum((df['first_ofd'] - df['rider_reached_store']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
df['verified']=pd.to_datetime(df['verified'])
df['booked']=pd.to_datetime(df['booked'])
```

```python
#Time taken for verification
df['Verified_in_Min']=np.where(df['verified'].notna() & df['booked'].notna(),
np.maximum((df['verified'] - df['booked']).dt.total_seconds() / 60, 0),np.nan)
df
```

```python
df['Return_Cancelled']=np.where(df['LATESTORDERSTATUS']=='RETURN_REQUEST_CANCELLED',1,0)
df
```
